```python
import numpy as np 
import matplotlib.pyplot as plt
```

DATA GENERATION -> 3 kluster with 2D normal distribution

```python
rng = np.random.default_rng(42)

# True GNN parameter
true_means = np.array([
    [0 , 0],
    [5 , 5],
    [-4, 5]
])

true_cov = np.array([
    [[1, 0.3] , [0.3, 1]],
    [[1.5, 0], [0, 0.8]],
    [[0.8, -0.2], [-0.2, 1.2]]
])

true_weights = np.array([0.3, 0.4, 0.3])

n = 600
K = 3

# Tirage des variable lattente 
Z = rng.choice(K, size=n, p=true_weights)

# Data generation 
X = np.array([
    rng.multivariate_normal(true_means[k], true_cov[k])
    for k in Z
    ])

plt.scatter(X[:,0], X[:,1], s=10)
plt.axis('equal')
plt.show()
```
